```python
# 1. Read the raw CSV data into PySpark DataFrames

orders_df = spark.read.csv("/Volumes/jet_catalog/default/jet_data_volume/raw_orders.csv", header=True, inferSchema=True)
weather_df = spark.read.csv("/Volumes/jet_catalog/default/jet_data_volume/raw_weather.csv", header=True, inferSchema=True)

# 2. Save the DataFrames as Bronze Delta Tables

orders_df.write.format("delta").mode("overwrite").saveAsTable("raw_orders_bronze")
weather_df.write.format("delta").mode("overwrite").saveAsTable("raw_weather_bronze")

# 3. Display the data to verify it loaded correctly
print("Orders Data:")
display(orders_df)

print("Weather Data:")
display(weather_df)
```
